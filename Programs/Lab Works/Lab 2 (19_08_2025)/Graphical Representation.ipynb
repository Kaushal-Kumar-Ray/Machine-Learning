```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load splits
train_df = pd.read_csv("train.csv")
val_df = pd.read_csv("val.csv")
test_df = pd.read_csv("test.csv")
```

```python
# Clean "Total Speed" → keep only numbers
for df in [train_df, val_df, test_df]:
    df["Total Speed"] = (
        df["Total Speed"].astype(str).str.extract(r"(\d+\.?\d*)")[0].astype(float)
    )
```

```python
# Plot distributions
plt.figure(figsize=(8,5))
plt.hist(train_df["Total Speed"].dropna(), bins=20, alpha=0.5, label="Train")
plt.hist(val_df["Total Speed"].dropna(), bins=20, alpha=0.5, label="Validation")
plt.hist(test_df["Total Speed"].dropna(), bins=20, alpha=0.5, label="Test")
plt.xlabel("Total Speed")
plt.ylabel("Count")
plt.title("Total Speed distribution in Train / Val / Test")
plt.legend()
plt.show()
```
